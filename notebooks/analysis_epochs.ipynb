```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# CAMBIA QUESTO PATH quando hai il file dal cluster
df = pd.read_csv("epochs_ALL.csv")

df.head()
```

```python
df_mean = df.groupby("epoch").mean()
df_mean
```

```python
plt.figure()

plt.plot(df_mean.index, df_mean["loss"], label="Total loss")
plt.plot(df_mean.index, df_mean["fixed_loss"], label="Fixed loss")
plt.plot(df_mean.index, df_mean["random_loss"], label="Random loss")

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Loss vs Epoch")
plt.legend()
plt.grid()

plt.show()
```

```python
plt.figure()

plt.plot(df_mean.index, df_mean["accuracy"], label="Accuracy")

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Accuracy vs Epoch")
plt.legend()
plt.grid()

plt.show()
```
