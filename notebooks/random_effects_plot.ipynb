```python
import numpy as np
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt
import ternary
```

```python
re = np.load("random_effects_ALL_....npy")  # cambia nome file
print(re.shape)
```

```python
def convert_to_probabilities(re):
    re_tensor = torch.tensor(re)
    re_prob = F.softmax(re_tensor, dim=1)
    return re_prob.numpy()

re_prob = convert_to_probabilities(re)
```

```python
def plot_ternary_paper_style(re_prob):

    scale = 100

    fig, tax = ternary.figure(scale=scale)
    fig.set_size_inches(6, 5)

    # punti
    points = re_prob * scale

    tax.scatter(
        points,
        marker='o',
        color='black',
        s=8,
        alpha=0.6
    )

    # griglia (come paper)
    tax.gridlines(
        multiple=10,
        color="gray",
        linewidth=0.5,
        linestyle="dotted"
    )

    # ticks
    tax.ticks(axis='lbr', multiple=20, linewidth=1)

    # labels (IMPORTANTE ordine)
    tax.left_axis_label("negative", fontsize=12)
    tax.right_axis_label("positive", fontsize=12)
    tax.top_axis_label("neutral", fontsize=12)

    # rimuove bordi inutili
    tax.clear_matplotlib_ticks()

    tax.set_title("Distribution of annotator biases", fontsize=14)

    tax.show()
```

```python
plot_ternary_paper_style(re_prob)
```
